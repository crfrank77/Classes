```julia
using JuMP
using HiGHS

model = Model(HiGHS.Optimizer)

# -------------------------
# Variables
# -------------------------
@variable(model, 2 <= x <= 8)
@variable(model, y >= 0)

# -------------------------
# Objective
# -------------------------
@objective(model, Max, 10x + 5y)

# -------------------------
# Constraints
# -------------------------
c1 = @constraint(model, 5x + 2y <= 50)
c2 = @constraint(model, x + 5y <= 30)

# -------------------------
# Solve
# -------------------------
optimize!(model)

println("\n===== Optimal Solution =====")
println("x* = ", value(x))
println("y* = ", value(y))
println("Objective = ", objective_value(model))

# -------------------------
# Duals for linear constraints
# -------------------------
println("\n===== Duals for Linear Constraints =====")
println("Dual of 5x + 2y ≤ 50: ", shadow_price(c1))
println("Dual of x + 5y ≤ 30: ", shadow_price(c2))

# -------------------------
# Duals for variable bounds
# -------------------------
println("\n===== Duals for Variable Bounds =====")
println("Dual of x ≥ 2: ", shadow_price(lower_bound(x)))
println("Dual of x ≤ 8: ", shadow_price(upper_bound(x)))
println("Dual of y ≥ 0: ", shadow_price(lower_bound(y)))
```
